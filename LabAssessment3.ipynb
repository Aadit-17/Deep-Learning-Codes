### Q1

1. Build a simple Recurrent Neural Network (RNN) using Keras to read a CSV file containing time series data and make a basic prediction?
 You can use a weather dataset from Dataset Source: https://github.com/velicki/Weather_Data_Analysis_Project/blob/main/Weather_Data.csv
 Normalize the data using MinMaxScaler from sklearn.
 Create sequences of data for the RNN to learn from.
 Use Keras to define a simple RNN model.
 Fit the model to your data.
 Use the trained model to make predictions.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN
from sklearn.metrics import mean_squared_error
```

```python
# Load dataset
url = "Weather_Data.csv"
df = pd.read_csv(url)
data = df['Temp_C'].values
```

```python
# Normalize data
scaler = MinMaxScaler()
data = scaler.fit_transform(data.reshape(-1, 1))
```

```python
# Create sequences
sequence_length = 10
X, y = [], []
for i in range(len(data) - sequence_length):
    X.append(data[i:i + sequence_length])
    y.append(data[i + sequence_length])
X, y = np.array(X), np.array(y)
```

```python
# Split into train and test sets
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]
```

```python
# Define RNN model
model = Sequential([
    SimpleRNN(50, activation='relu', input_shape=(sequence_length, 1)),
    Dense(1)
])
model.compile(optimizer='adam', loss='mse')
```

```python
model.summary()
```

```python
model.fit(X_train, y_train, epochs=20, batch_size=16)
```

```python
# Predict
predictions = model.predict(X_test)
```

```python
np.sqrt(mean_squared_error(y_test, predictions))
```

### Q2

2.Build a simple Gated Recurrent Unit (GRU) model using Keras to perform sentiment analysis on the IMDB dataset?
 Use Keras to load the IMDB dataset.
 Define a simple GRU model using Keras.
 Fit the model to the training data.
 Evaluate the model on the test data.
 Use the trained model to make predictions on new data.

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Embedding
```

```python
# Load IMDB dataset
max_features = 10000  # Number of words to consider as features
maxlen = 100  # Cut texts after this many words
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=max_features)
```

```python
# Preprocess data
X_train = pad_sequences(X_train, maxlen=maxlen)
X_test = pad_sequences(X_test, maxlen=maxlen)
```

```python
# Define GRU model
model = Sequential([
    Embedding(max_features, 128, input_length=maxlen),
    GRU(64, dropout=0.2, recurrent_dropout=0.2),
    Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Evaluate
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test Accuracy:", test_acc)
```

3.Build a simple Long Short-Term Memory (LSTM) model using Keras to predict masked words in a sentence?
Sample text data
texts = [
"I love to [MASK] programming.",
"This is a [MASK] example.",
"Keras makes building models [MASK]."
]
Example prediction
masked_sentence = "I love to [MASK] programming."

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
texts = [
    "I love to [MASK] programming.",
    "This is a [MASK] example.",
    "Keras makes building models [MASK]."
]
```

```python
# Preprocess text
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)
vocab_size = len(tokenizer.word_index) + 1
```

```python
# Pad sequences
max_len = max(len(seq) for seq in sequences)
sequences = pad_sequences(sequences, maxlen=max_len)
```

```python
# Define LSTM model
model = Sequential([
    Embedding(vocab_size, 128, input_length=max_len),
    LSTM(64, return_sequences=False),
    Dense(vocab_size, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
```

```python
model.summary()
```
